```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("https://raw.githubusercontent.com/RajeshRanaGiet/DMPM-Lab/main/Experiment%204/line%20chart.csv")
data = data.iloc[:,1:]
data.plot()
plt.title("Line Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.xticks()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("https://raw.githubusercontent.com/RajeshRanaGiet/DMPM-Lab/main/Experiment%204/line%20chart.csv")
sns.barplot(x=data.index, y=data['A'])
plt.title("Bar PLot")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("https://raw.githubusercontent.com/RajeshRanaGiet/DMPM-Lab/main/Experiment%204/line%20chart.csv")
y = data['A']
plt.scatter(x = data.index, y=y)
plt.title("Scatter Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.xticks()
plt.show()
```

```python
import matplotlib.pyplot as plt
cars = ['AUDI', 'BMW', 'FORD','TESLA', 'JAGUAR',]
data = [23, 10, 35, 15, 10]
plt.pie(data, labels=cars, autopct='%1.1f%%')
plt.title(" Pie Chart")
plt.show()
```
