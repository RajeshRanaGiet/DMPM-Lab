```python
import numpy as np
```

```python
#Q1
l = [int(x) for x in input("Enter the list of numer :").split(",")]
mean_without = sum(l)/len(l)
print(f"The mean is {mean_without}")
```

```python
#Q2
l = [int(x) for x in input("Enter the list of numer :").split(",")]
mean_with = np.mean(l)
print(f"The mean is {mean_with}")
```

```python
#Q3
l = [int(x) for x in input("Enter the list of numer :").split(",")]
if len(l)%2 == 0:
    median_without = (l[len(l)//2] + l[len(l)//2 - 1])/2
else:
    median_without = l[len(l)//2]
print(f"The median is {median_without}")
```

```python
#Q4
l = [int(x) for x in input("Enter the list of numer :").split(",")]
median_with = np.median(l)
print(f"The median is {median_with}")
```

```python
#Q5
l = [int(x) for x in input("Enter the list of numer :").split(",")]
dict = {}
for x in l:
    if x in dict:
        dict[x] += 1
    else:
        dict[x] = 1
mode_without = max(dict, key=dict.get)
print(mode_without)
```

```python
#Q6
l = [int(x) for x in input("Enter the list of numer :").split(",")]
mode_with = np.bincount(l).argmax()
print(mode_with)
```
