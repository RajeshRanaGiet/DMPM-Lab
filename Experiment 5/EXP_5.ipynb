```python
#Q1
import numpy as np
l = np.array([10,20,30,40,50])

#z-score normalization
mean = l.mean()
sd= l.std()
print("The z-score normalization for : ")
for i in l:
  print(f"{i} is {(i-mean)/sd:.4f}")

#min-max normalization
min = l.min()
max = l.max()
print("\nThe min-max normalization for : ")
for i in l:
  print(f"{i} is {(i-min)/(max-min):.4f}")

#Decimal Scalling Normalization
j = len(str(max))
print("\nThe decimal scalling normalization for : ")
for i in l:
  print(f"{i} is {i/(10**j):.4f}")
```
